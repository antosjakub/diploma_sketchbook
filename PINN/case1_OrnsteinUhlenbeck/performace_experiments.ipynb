```python
import torch
```

```python
def gelu(x):
    return x * 0.5 * (1.0 + torch.erf(x / 1.41421))

@torch.compile
def gelu_com(x):
    return x * 0.5 * (1.0 + torch.erf(x / 1.41421))
```

```python
X = torch.rand(100_000_000)
```

```python
gelu(X)
```

```python
gelu_com(torch.rand(10))
```

```python
gelu_com(X)
```
