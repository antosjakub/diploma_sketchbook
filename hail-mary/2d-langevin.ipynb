# Langevin model of energy-cycling active particles in 2D

We track $N$ identical particles in a 2D domain $\Omega \subset \mathbb{R}^2$. Each particle $i = 1, \ldots, N$ carries

- a **position** $\mathbf{x}_i(t) \in \Omega$,
- an **internal energy** $e_i(t) \in [0, e_{\max}]$.

Two distinguished sites:
- charging station $\mathbf{x}_A = (0, 0)$,
- discharging station $\mathbf{x}_B \in \Omega$, $\mathbf{x}_B \neq \mathbf{x}_A$.

The qualitative dynamics we want to reproduce:

- a *low-energy* particle is drawn toward $A$ to refuel,
- a *high-energy* particle is drawn toward $B$ to deliver,
- the role-switch happens smoothly at some threshold energy $e^\star$.

Below we write the **microscopic SDE** (the system that will be simulated), then the **macroscopic PDE** (Fokker–Planck) that its empirical distribution satisfies in the mean-field limit, then fix the constitutive choices, observables and boundary conditions.


## 1. Microscopic model — Langevin SDE

For each $i = 1, \ldots, N$,

$$
\boxed{\;
\begin{aligned}
d\mathbf{x}_i &= \mathbf{v}\bigl(\mathbf{x}_i,\, e_i\bigr)\,dt \;+\; \sqrt{2\,D}\;d\mathbf{W}_i^{(x)}, \\[4pt]
d e_i        &= g\bigl(\mathbf{x}_i,\, e_i\bigr)\,dt \;+\; \sqrt{2\,\tilde D}\;d W_i^{(e)},
\end{aligned}
\;}
$$

where $\mathbf{W}_i^{(x)} \in \mathbb{R}^2$ and $W_i^{(e)} \in \mathbb{R}$ are independent standard Brownian motions, $D > 0$ is the spatial diffusivity, and $\tilde D \geq 0$ is the energy diffusivity (small; models stochasticity of charging / consumption).

Two ingredients to specify:
- the **state-dependent drift** $\mathbf{v}(\mathbf{x}, e)$, which encodes "head toward $A$ if low, toward $B$ if full",
- the **local energy flow** $g(\mathbf{x}, e)$, which encodes "charge near $A$, discharge near $B$, leak everywhere".

Both are given concrete forms in §3.


## 2. Macroscopic model — Fokker–Planck PDE

Let $p(\mathbf{x}, e, t)$ be the joint density of a single particle (equivalently, the mean-field limit of the empirical distribution). The Itô SDE above corresponds, via the **Kolmogorov forward equation**, to

$$
\boxed{\;
\partial_t p \;=\; -\,\nabla_{\mathbf{x}} \!\cdot\!\bigl[\mathbf{v}(\mathbf{x},e)\,p\bigr] \;+\; D\,\Delta_{\mathbf{x}}\, p \;-\; \partial_e\!\bigl[g(\mathbf{x},e)\,p\bigr] \;+\; \tilde D\,\partial_e^2 p .
\;}
$$

In flux form $\partial_t p + \nabla \cdot \mathbf{J} = 0$, where $\nabla = (\nabla_{\mathbf{x}}, \partial_e)$ acts on the **joint state space** $\Omega \times [0, e_{\max}]$, the probability current is

$$
\mathbf{J}(\mathbf{x},e,t) \;=\;
\begin{pmatrix}
\mathbf{J}_{\mathbf{x}} \\[2pt] J_e
\end{pmatrix}
\;=\;
\begin{pmatrix}
\mathbf{v}(\mathbf{x},e)\,p \;-\; D\,\nabla_{\mathbf{x}} p \\[2pt]
g(\mathbf{x},e)\,p \;-\; \tilde D\,\partial_e p
\end{pmatrix}.
$$

The connection between the SDE and the PDE is the **empirical measure**

$$\hat p_N(\mathbf{x}, e, t) \;:=\; \frac{1}{N}\sum_{i=1}^{N}\, \delta\!\bigl(\mathbf{x} - \mathbf{x}_i(t)\bigr)\,\delta\!\bigl(e - e_i(t)\bigr),$$

which converges weakly to $p$ as $N \to \infty$ (independent particles, no interactions). For a single particle ($N = 1$), $p$ is exactly the law of $(\mathbf{x}_1, e_1)(t)$.


## 3. Constitutive choices

### 3.1 Switching function

Let $h : [0, e_{\max}] \to [0, 1]$ smoothly interpolate from "uncharged" ($h = 0$) to "charged" ($h = 1$),

$$h(e) \;=\; \tfrac{1}{2}\Bigl[\,1 + \tanh\!\bigl(\tfrac{e - e^\star}{\Delta e}\bigr)\Bigr],$$

with crossover energy $e^\star$ and width $\Delta e$.

### 3.2 Attractive potentials

Quadratic wells centred at the two sites,

$$U_A(\mathbf{x}) = \tfrac{1}{2}\,k_A\,|\mathbf{x} - \mathbf{x}_A|^2, \qquad U_B(\mathbf{x}) = \tfrac{1}{2}\,k_B\,|\mathbf{x} - \mathbf{x}_B|^2,$$

with stiffnesses $k_A, k_B > 0$. The corresponding pulls are linear springs,

$$-\nabla U_A(\mathbf{x}) = -k_A\,(\mathbf{x} - \mathbf{x}_A), \qquad -\nabla U_B(\mathbf{x}) = -k_B\,(\mathbf{x} - \mathbf{x}_B).$$

### 3.3 State-dependent drift

Blend the two pulls with $h(e)$:

$$\mathbf{v}(\mathbf{x}, e) \;=\; -\bigl(1 - h(e)\bigr)\,\nabla U_A(\mathbf{x}) \;-\; h(e)\,\nabla U_B(\mathbf{x}).$$

So $\mathbf{v}(\mathbf{x}, e \to 0) \to -k_A(\mathbf{x} - \mathbf{x}_A)$ (pure pull toward $A$), and $\mathbf{v}(\mathbf{x}, e \to e_{\max}) \to -k_B(\mathbf{x} - \mathbf{x}_B)$ (pure pull toward $B$).

### 3.4 Localized energy exchange

Normalized Gaussian "footprints" around each site, of size $\ell$:

$$\chi_A(\mathbf{x}) = \frac{1}{2\pi\,\ell^2}\,\exp\!\Bigl(-\tfrac{|\mathbf{x} - \mathbf{x}_A|^2}{2\,\ell^2}\Bigr), \qquad \chi_B(\mathbf{x}) = \frac{1}{2\pi\,\ell^2}\,\exp\!\Bigl(-\tfrac{|\mathbf{x} - \mathbf{x}_B|^2}{2\,\ell^2}\Bigr).$$

The local energy flow combines charging at $A$, discharging at $B$, and uniform leakage:

$$
g(\mathbf{x}, e) \;=\; \underbrace{\gamma_A\,\chi_A(\mathbf{x})\,(e_{\max} - e)}_{\text{charging near }A} \;-\; \underbrace{\gamma_B\,\chi_B(\mathbf{x})\,e}_{\text{discharging near }B} \;-\; \underbrace{\lambda\,e}_{\text{leakage everywhere}}.
$$

The saturation factor $(e_{\max} - e)$ ensures charging stops at full battery; the factor $e$ in the discharge term ensures an empty particle cannot be drained further; the leakage term penalizes idle hoarding.


## 4. Conservation laws and observables

### 4.1 Particle number

Integrating the PDE over $\Omega \times [0, e_{\max}]$ and using no-flux boundary conditions (§5) gives

$$\frac{d}{dt}\!\int p \,d\mathbf{x}\,de = 0,$$

so total particle number is exactly conserved (the SDE creates / destroys nothing).

### 4.2 Marginal densities

The **spatial density** and **energy-weighted density** are

$$n(\mathbf{x}, t) := \int_0^{e_{\max}}\! p(\mathbf{x}, e, t)\,de, \qquad \mathcal{E}(\mathbf{x}, t) := \int_0^{e_{\max}}\! e\,p(\mathbf{x}, e, t)\,de.$$

Multiplying the PDE by $1$ and by $e$ and integrating in $e$ yields transport equations

$$\partial_t n + \nabla_{\mathbf{x}} \cdot \langle \mathbf{v}\rangle = D\,\Delta_{\mathbf{x}} n,$$

$$\partial_t \mathcal{E} + \nabla_{\mathbf{x}} \cdot \langle e\,\mathbf{v}\rangle = D\,\Delta_{\mathbf{x}}\,\mathcal{E} + \langle g\rangle,$$

with the $e$-moments $\langle f\rangle(\mathbf{x}, t) := \int f(\mathbf{x}, e)\,p(\mathbf{x}, e, t)\,de$. The energy density is **not** conserved — $\langle g\rangle$ injects power at $A$ and absorbs it at $B$.

### 4.3 The cycle current

The cycle current is the central observable. Pick any oriented curve $\Sigma \subset \Omega$ that separates $A$ from $B$, with unit normal $\hat{\mathbf{n}}_\Sigma$ pointing from the $A$-side to the $B$-side. Define the **charged-particle current across $\Sigma$**:

$$J_{AB} \;:=\; \int_\Sigma \!\int_0^{e_{\max}} h(e)\;\mathbf{J}_{\mathbf{x}}(\mathbf{x}, e, t)\cdot\hat{\mathbf{n}}_\Sigma \;\;de\,d\ell .$$

At steady state, $\partial_t p = 0$ and $\nabla \cdot \mathbf{J} = 0$, so $J_{AB}$ is **independent of which separating curve $\Sigma$** is chosen — this is the global cycle current. The analogous "uncharged return current" using $1 - h(e)$ has the opposite sign, and the two cancel in the total $n$-current (since total mass is conserved).

### 4.4 Energy throughput

Splitting $g = g_+ - g_-$ into positive and negative parts, the steady-state input and dissipation rates are

$$\dot Q_{\rm in} = \int g_+(\mathbf{x}, e)\, p \,d\mathbf{x}\,de, \qquad \dot Q_{\rm out} = \int g_-(\mathbf{x}, e)\, p \,d\mathbf{x}\,de,$$

and $\dot Q_{\rm in} = \dot Q_{\rm out}$ at steady state (input at $A$ balances output at $B$ plus leakage everywhere).


## 5. Boundary conditions

### 5.1 Spatial boundary $\partial\Omega$

Three reasonable choices, all of which keep total particle number conserved:

1. **Reflective walls**: $\mathbf{J}_{\mathbf{x}} \cdot \hat{\mathbf{n}}\big|_{\partial\Omega} = 0$. For the SDE, reflect particle positions when they cross a wall.
2. **Periodic** (torus $\mathbb{T}^2$). Wrap positions modulo the period.
3. **Unbounded** $\Omega = \mathbb{R}^2$ with $p \to 0$ at infinity. Works because both wells are confining.

For numerical work the simplest is a rectangular domain of side $\sim 2 L$ (where $L = |\mathbf{x}_B - \mathbf{x}_A|$) with reflective walls — large enough that the walls don't interfere with the cycle.

### 5.2 Energy boundary

At $e = 0$ and $e = e_{\max}$: **reflective**, $J_e\big|_{e \in \{0, e_{\max}\}} = 0$. The saturation factors in $g$ already drive the energy *drift* to zero at the boundaries; reflective BCs handle the energy *diffusion*. In the SDE, reflect any energy excursion: replace $e_i \to |e_i|$ at the low end, $e_i \to 2\,e_{\max} - e_i$ at the high end (or simply clip).


## 6. Parameter glossary

| Symbol | Meaning | Typical role |
|---|---|---|
| $\mathbf{x}_A,\, \mathbf{x}_B$ | site locations | fixed geometry |
| $L = \lvert\mathbf{x}_B - \mathbf{x}_A\rvert$ | inter-site distance | length scale |
| $k_A,\, k_B$ | well stiffnesses | how hard each site pulls |
| $D$ | spatial diffusivity | thermal noise on motion |
| $\tilde D$ | energy diffusivity | noise on charging/consumption |
| $e_{\max}$ | full battery | upper energy bound |
| $e^\star,\, \Delta e$ | switching threshold & width | "I have enough fuel to deliver" |
| $\gamma_A,\, \gamma_B$ | charging / discharging rates | how fast the stations work |
| $\ell$ | station footprint radius | spatial extent of exchange |
| $\lambda$ | leakage rate | passive energy loss |
| $N$ | number of particles | $\hat p_N \to p$ as $N \to \infty$ |

Three dimensionless groups will govern the qualitative behaviour:

- **Péclet** ${\rm Pe} = k_A L^2 / D$ — drift vs. spatial diffusion (need $\gg 1$ so particles actually reach their target).
- **Charging Damköhler** ${\rm Da}_A = \gamma_A / k_A$ — charging speed vs. arrival speed (need $\gtrsim 1$ so charging completes during a visit).
- **Leakage** $\tilde\lambda = \lambda L^2 / D$ — leakage vs. transit time (need $\ll 1$ so the bus arrives still loaded).
