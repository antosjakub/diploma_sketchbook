```python
import torch
```

```python
size = '10k'
d=2
data = torch.load(f'testing_data__d={d}_{size}.pth')['data']
X_pde = data['X_pde']
X_ic = data['X_ic'][:,:-1]
#X = torch.cat((X_pde, X_ic), dim=0)
#X.shape
X = X_pde
d = X.shape[-1]-1
X_pde[:,-1].max(), X.shape, d
```

```python
x = X[X[:, -1].argsort()]   # sort by time
n_parts = 4
parts = torch.chunk(x, n_parts, dim=0)        # roughly equal number of rows

bins = []
times = []
for i in range(len(parts)):
    times.append( parts[i][:,-1].max().item() )
    bins.append( parts[i][:,:-1].numpy() )
times
```

```python
bins
```

```python
import numpy as np
np.savez_compressed(f'X_pde_bins_d={d}_{size}.npz', X_pde_bins=np.array(bins), times=np.array(times))
np.savez_compressed(f'X_ic_d={d}_{size}.npz', X_ic=X_ic.numpy())
```
